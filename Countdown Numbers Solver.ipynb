```python
import random
```

```python
big = list(range(25, 101, 25))
small = list(range(1, 11)) * 2
```

```python
target = random.randint(100, 999)
```

```python
class Stage:
    def __init__(self, value, target):
        self.result_number = None
        self.value = value
        self.target = target
        self.away = target - value
        self.used = [False] * 6
        self.input_1 = None
        self.input_2 = None
        self.input_operation = None
        
    def use(self, place):
        self.used[place] = True
```

```python
# Pick the Numbers

def pick_numbers(big_number_count):
    big_number_count = 2

    x = random.sample(big, big_number_count)
    y = random.sample(small, 6 - big_number_count)

    return x + y

def overlapped(stage_1, stage_2):
    return (stage_1.used[0] and stage_2.used[0]
        or  stage_1.used[1] and stage_2.used[1]
        or  stage_1.used[2] and stage_2.used[2]
        or  stage_1.used[3] and stage_2.used[3]
        or  stage_1.used[4] and stage_2.used[4]
        or  stage_1.used[5] and stage_2.used[5])

def calculate(x, y, operation):
    match operation:
        case "Addition":
            return x + y
        
        case "Subtraction":
            return max(x, y) - min(x, y)
        
        case "Multiplication":
            return x * y
        
        case "Division":
            return max(x, y) / min(x, y)
        
```

```python
big_numbers = 2
numbers = pick_numbers(big_numbers)
```

```python
#Starting Stages

stages = []

for rank, n in enumerate(numbers):
    s = Stage(n, target)
    s.use(rank)
    s.result_number = rank
    stages.append(s)
```

```python
for s in stages:
    print(vars(s))
    print()
```

```python
for i in stages:
    for j in stages:
        if i.result_number < j.result_number and not overlapped(i, j):
            
            
            for m in ('Addition', 'Subtraction', 'Multiplication', 'Division'):
                operation_result = calculate(i.value, j.value, m)
                
                if operation_result > 0 and operation_result % 1 == 0:
                    new_stage = Stage(operation_result, target)
                    new_stage.result_number = len(stages)
                    new_stage.input_1 = i.result_number
                    new_stage.input_operation = m
                    new_stage.input_2 = j.result_number
                    
                    for u in (i, j):
                        for n, b in enumerate(u.used):
                            if b:
                                new_stage.use(n)
                    
                    stages.append(new_stage)
```
